```python


import ipywidgets as widgets
from IPython.display import display

# Membuat input nama pemain
input_nama = widgets.Text(
    value='',
    placeholder='Ketik nama pemain',
    description='Nama:',
    disabled=False
)

# Membuat pilihan level menggunakan slider
input_level = widgets.IntSlider(
    min=1,
    max=100,
    description='Level:',
    value=1
)

# Membuat pilihan role menggunakan dropdown
input_role = widgets.Dropdown(
    options=['jungle', 'mage', 'archer'],
    description='Role:'
)

# Membuat tombol daftar
tombol_daftar = widgets.Button(
    description='Daftar',
    button_style='primary'
)

# Tempat menampilkan hasil
output_hasil = widgets.Output()

# Fungsi saat tombol diklik
def daftar_klik(b):
    with output_hasil:
        output_hasil.clear_output()
        print(
            f'Pemain {input_nama.value}, '
            f'dengan role {input_role.value}, '
            f'level {input_level.value}, '
            f'telah siap bertarung!'
        )

# Menghubungkan tombol dengan fungsi
tombol_daftar.on_click(daftar_klik)

# Menampilkan semua widget
display(
    input_nama,
    input_level,
    input_role,
    tombol_daftar,
    output_hasil
)
```
